# 청크 크기 × 모델 검색 평가
각 실험의 `retrieval_eval.py` 결과를 모아 비교하고 top-k 원문을 검수합니다.

```python
from pathlib import Path
import pandas as pd
ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / 'plan.md').exists())
EVAL_DIR = ROOT / 'reports/legal_api_v2/eval'
```

```python
leaderboard = pd.read_csv(EVAL_DIR / 'leaderboard.csv')
leaderboard.sort_values(['recall_at_5', 'ndcg_at_10', 'latency_p95_ms'], ascending=[False, False, True])
```

```python
import json
experiment = leaderboard.sort_values(['recall_at_5', 'ndcg_at_10'], ascending=False).iloc[0]['experiment_id']
review_rows = [json.loads(line) for line in (EVAL_DIR / experiment / 'topk_review.jsonl').read_text(encoding='utf-8').splitlines()]
pd.json_normalize(review_rows, record_path='results', meta=[['question','question_id'], ['question','question']]).head(20)
```

우승 기준: Recall@5 → nDCG@10 → 청크 수·p95 지연시간. `questions.jsonl`의 gold ID를 팀이 최종 검수한 뒤 확정합니다.
