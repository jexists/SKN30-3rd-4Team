원본을 수정하지 않고 보고서 CSV와 그래프를 생성합니다.

```python
from pathlib import Path
import sys
ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / 'plan.md').exists())
sys.path.insert(0, str(ROOT))
from pipeline.analysis.prechunk_eda import analyze
from pipeline.common import load_pipeline_config, resolve_repo_path
```

```python
CONFIG = load_pipeline_config()
INPUT_DIR = resolve_repo_path(CONFIG['paths']['final_dir'])
OUTPUT_DIR = resolve_repo_path(CONFIG['paths']['reports_dir']) / 'prechunk_eda'
LENGTH_THRESHOLDS = (20, 300, 500, 800, 1000, 1500, 2000)
CHUNK_CANDIDATES = tuple(CONFIG['chunking']['candidates'])
OVERLAP_RATIO = float(CONFIG['chunking']['overlap_ratio'])
```

```python
report = analyze(
    INPUT_DIR, OUTPUT_DIR, with_tokenizers=False,
    length_thresholds=LENGTH_THRESHOLDS,
    chunk_candidates=CHUNK_CANDIDATES,
    overlap_ratio=OVERLAP_RATIO,
)
report
```

모델 파일을 내려받을 수 있는 환경에서는 `with_tokenizers=True`로 다시 실행해 모델별 8,192 토큰 초과 여부도 생성합니다.

```python
print(f"추천 기준선: {report['recommended_chunk_size']}자 / overlap {report['recommended_overlap']}자")
print(report['recommendation_reason'])
```
